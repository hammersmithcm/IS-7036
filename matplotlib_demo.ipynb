```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]


plt.plot(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Plot')
plt.show()
```

```python
plt.scatter(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot')
plt.show()
```

```python
#changing parts of graph: like color and fontsize
plt.scatter(x, y, c ="green", edgecolor ="black")
plt.xlabel('X-axis', fontsize =14)
plt.ylabel('Y-axis', fontsize =14)
plt.title('Scatter Plot', fontsize = 18, weight='bold')
plt.show()
```

```python
titanic =pd.read_csv("titanic_clean.csv")
titanic.head(10)
```

```python
titanic['age'].dtype
```

```python
age_data = titanic['age'].dropna()
```

```python
#to determine bin width
max_val = np.max(age_data)
max_val
```

```python
plt.figure(figsize=(8, 6))
plt.hist(age_data, bins=8, color="darksalmon", edgecolor="white", alpha=0.7)
plt.title("Age Distribution of Titanic passengers", fontsize=17, weight="bold")
plt.xlabel("Age", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```
